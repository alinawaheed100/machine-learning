```python
import numpy as np

# Make a copy for cleaning
df_clean = df.copy()

# 1. Clean 'Total Fat (g)' - remove potential non-numeric values and convert to float
df_clean['Total Fat (g)'] = pd.to_numeric(df_clean['Total Fat (g)'].astype(str).str.replace(r'[^0-9.]', '', regex=True), errors='coerce')

# 2. Clean '% DV' columns - remove '%' sign and convert to float
dv_cols = ['Vitamin A (% DV)', 'Vitamin C (% DV)', 'Calcium (% DV)', 'Iron (% DV)']
for col in dv_cols:
    df_clean[col] = pd.to_numeric(df_clean[col].astype(str).str.replace('%', ''), errors='coerce')

# 3. Clean 'Caffeine (mg)' - replace values like 'Varies' or 'varies' with NaN, then convert
df_clean['Caffeine (mg)'] = pd.to_numeric(df_clean['Caffeine (mg)'].astype(str).str.replace('Varies', 'NaN', case=False), errors='coerce')

# Fill missing values created during conversions with the median of their respective columns
num_cols_to_fill = ['Total Fat (g)', 'Caffeine (mg)'] + dv_cols
for col in num_cols_to_fill:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

print("=== Cleaned Data Types and Missing Values ===")
print(df_clean[num_cols_to_fill].dtypes)
print("\nAny missing values left?", df_clean.isnull().any().any())
```

```python
from sklearn.model_selection import train_test_split

# Splitting the data: 80% training and 20% testing
train_df, test_df = train_test_split(df_clean, test_size=0.2, random_state=42)

print(f"Training set shape: {train_df.shape}")
print(f"Testing set shape: {test_df.shape}")
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, accuracy_score
import matplotlib.pyplot as plt
from sklearn import tree

# Define the features (X) and the target (y)
features = ['Calories', 'Total Fat (g)', 'Trans Fat (g)', 'Saturated Fat (g)', 'Sodium (mg)', 'Total Carbohydrates (g)', 'Cholesterol (mg)', 'Dietary Fibre (g)', 'Sugars (g)', 'Protein (g)', 'Caffeine (mg)']
target = 'Beverage_category'

X_train = train_df[features]
y_train = train_df[target]
X_test = test_df[features]
y_test = test_df[target]

# Initialize and train the Decision Tree Classifier
clf = DecisionTreeClassifier(max_depth=5, random_state=42)
clf.fit(X_train, y_train)

# Predict on the test set
y_pred = clf.predict(X_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy * 100:.2f}%")
print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
# Visualize the Decision Tree structure
plt.figure(figsize=(20, 10))
tree.plot_tree(
    clf,
    feature_names=features,
    class_names=clf.classes_,
    filled=True,
    rounded=True,
    fontsize=10
)
plt.title("Visualized Decision Tree for Starbucks Beverage Category Classification", fontsize=16)
plt.show()
```

```python
# Take 10 samples from the test data
sample_test = test_df.head(10).copy()

# Extract the features for these 10 samples
X_sample = sample_test[features]

# Generate predictions
sample_test['Predicted_category'] = clf.predict(X_sample)

# Display the actual vs predicted labels along with the beverage name
comparison_df = sample_test[['Beverage', 'Beverage_prep', 'Beverage_category', 'Predicted_category']]
comparison_df.columns = ['Beverage Name', 'Prep', 'Actual Category', 'Predicted Category']
display(comparison_df)
```
